```python
import matplotlib.pyplot as plt
yield_by_season = df.groupby("Season")["Yield_Tonnes_Ha"].mean()
yield_by_season.plot(kind="bar")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.title("Average Yield by Season")
plt.show()
```

```python
# Average profit by Season
```

```python
profit_by_season = df.groupby("Season")["Profit_INR"].mean()

profit_by_season.plot(kind="bar")
plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Total production by season.
```

```python
production_by_season = df.groupby("Season")["Production_Tonnes"].sum()

production_by_season.plot(kind="bar")
plt.title("Total Production by Season")
plt.xlabel("Season")
plt.ylabel("Total Production (Tonnes)")
plt.xticks(rotation=0)
plt.show()
```
